```python
import pandas as pd
import googlemaps
import time
import os

API_KEY = "TU_API_KEY"
gmaps = googlemaps.Client(key=API_KEY)

archivo_cache = "distancias_rutas_google.csv"

rutas = df[['ORIGEN', 'DESTINO', 'National']].drop_duplicates().copy()

print("Rutas únicas:", len(rutas))

def construir_direccion(lugar, national):
    lugar = str(lugar)

    if national == True or str(national).upper() in ['TRUE', 'YES', 'SI', 'SÍ', 'NACIONAL']:
        direccion = lugar + ", Mexico"
    else:
        direccion = lugar

    return direccion


if os.path.exists(archivo_cache):

    rutas_distancias = pd.read_csv(archivo_cache)

else:

    distancias_km = []
    duraciones_min = []

    for i in range(len(rutas)):

        origen = construir_direccion(
            rutas.iloc[i]['ORIGEN'],
            rutas.iloc[i]['National']
        )

        destino = construir_direccion(
            rutas.iloc[i]['DESTINO'],
            rutas.iloc[i]['National']
        )

        resultado = gmaps.distance_matrix(
            origins=[origen],
            destinations=[destino],
            mode="driving",
            units="metric",
            language="es"
        )

        status_general = resultado['status']
        status_ruta = resultado['rows'][0]['elements'][0]['status']

        if status_general == "OK" and status_ruta == "OK":
            distancia_metros = resultado['rows'][0]['elements'][0]['distance']['value']
            duracion_segundos = resultado['rows'][0]['elements'][0]['duration']['value']

            distancias_km.append(distancia_metros / 1000)
            duraciones_min.append(duracion_segundos / 60)

        else:
            distancias_km.append(None)
            duraciones_min.append(None)

        time.sleep(0.1)

    rutas['distancia_km'] = distancias_km
    rutas['duracion_min'] = duraciones_min

    rutas.to_csv(archivo_cache, index=False)
    rutas_distancias = rutas.copy()

df = df.merge(
    rutas_distancias,
    on=['ORIGEN', 'DESTINO', 'National'],
    how='left'
)
```

```python
# convertir df en un csv
df.to_csv('../Data/TEC Loadboard - BD V1 Limpio.csv', index=False)
```
